```python
# 387. First Unique Character in a String

class Solution:
    def firstUniqChar(self, s: str) -> int:
        hash={}
        for i in s:
            if i not in hash:
                hash[i]=1
            else:
                hash[i]+=1
        for i in range(len(s)):
            if hash[s[i]]==1:
                return i
        return -1
```

```python
# 383. Ransom Note

class Solution:
    def canConstruct(self, ransomNote: str, magazine: str) -> bool:
        hash_r={}
        hash_m={}
        for i in ransomNote:
            if i not in hash_r:
                hash_r[i]=1
            else:
                hash_r[i]+=1
        for i in magazine:
            if i not in hash_m:
                hash_m[i]=1
            else:
                hash_m[i]+=1
        for i in hash_r:
            if i not in hash_m:
                return False
            if hash_r[i]>hash_m[i]:
                return False
        return True
```
