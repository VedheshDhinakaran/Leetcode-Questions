```python
# 28. Find the first occurrence of a substring in a string
class Solution:
    def strStr(self, haystack: str, needle: str) -> int:
        n=len(needle)
        for i in range(len(haystack)):
            if haystack[i:i+n]==needle:
                return i
        return -1
```

```python
# 58. Length of Last Word
class Solution:
    def lengthOfLastWord(self, s: str) -> int:
        res=s.split()
        return len(res[-1])
```

```python
# 125. Valid Palindrome
class Solution:
    def isPalindrome(self, s: str) -> bool:
        res=""
        for i in s:
            if i.isalpha() or i.isdigit():
                res+=i.lower()
        if res==res[::-1]:
            return True
        return False
```

```python
# 67. Add Binary

class Solution:
    def binToDec(self,b):
        res=0
        p=0
        for i in range(len(b)-1,-1,-1):
            res+=(int(b[i])*(2**p))
            p+=1
        return res
    def decToBin(self,d):
        if d==0:
            return "0"
        res=""
        while d>0:
            res+=str(d%2)
            d=d//2
        return res[::-1]
    def addBinary(self, a: str, b: str) -> str:
        n1=self.binToDec(a)
        n2=self.binToDec(b)
        res=n1+n2
        ans=self.decToBin(res)
        return ans
```

```python
# 202. Happy Number

class Solution:
    def func(self,n,seen):
        if n==1:
            return True
        if n in seen:
            return False
        seen.append(n)
        arr=[int(x) for x in str(n)]
        res=0
        for i in arr:
            res+=(i*i)
        return self.func(res,seen)
    def isHappy(self, n: int) -> bool:
        return self.func(n,[])
```

```python
# 171. Excel Sheet Column Number

class Solution:
    def titleToNumber(self, columnTitle: str) -> int:
        res=0
        for i in columnTitle:
            res=res*26+(ord(i)-ord("A")+1)
        return res
```

```python
# 168. Excel Sheet Column Title

class Solution:
    def convertToTitle(self, columnNumber: int) -> str:
        res=""
        while columnNumber>0:
            columnNumber-=1
            res+=chr((columnNumber%26)+ord("A"))
            columnNumber=columnNumber//26
        return res[::-1]
```

```python
# 389. Find the Difference

class Solution:
    def findTheDifference(self, s: str, t: str) -> str:
        ans=0
        for c in s:
            ans^=ord(c)
        for c in t:
            ans^=ord(c)
        return chr(ans)
```
