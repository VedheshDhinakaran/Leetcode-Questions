```python
# 344. Reverse String

class Solution:
    def reverseString(self, s: List[str]) -> None:
        """
        Do not return anything, modify s in-place instead.
        """
        n=len(s)
        for i in range(n//2):
            s[i],s[n-1-i]=s[n-1-i],s[i]
        return s
```

```python
# 345. Reverse Vowels of a String

class Solution:
    def reverseVowels(self, s: str) -> str:
        arr=list(s)
        vowels=[]
        for i in arr:
            if i in "aeiouAEIOU":
                vowels.append(i)
        n=len(vowels)
        for i in range(n//2):
            vowels[i],vowels[n-i-1]=vowels[n-i-1],vowels[i]
        j=0
        res=""
        for i in range(len(arr)):
            if arr[i] not in "aeiouAEIOU":
                res+=arr[i]
            else:
                res+=vowels[j]
                j+=1
        return res
```

```python
# 541. Reverse String II

class Solution:
    def reverseStr(self, s: str, k: int) -> str:
        if len(s)<k:
            return s[::-1]
        res=""
        i=0
        while i<len(s):
            res+=s[i:i+k][::-1]
            res+=s[i+k:i+2*k]
            i+=2*k
        return res
```

```python
# 392. Is Subsequence

class Solution:
    def isSubsequence(self, s: str, t: str) -> bool:
        n1=len(s)
        n2=len(t)
        c=0
        i=0
        j=0
        while i<n1 and j<n2:
            if s[i]==t[j]:
                c+=1
                j+=1
                i+=1
            else:
                j+=1
        if c==n1:
            return True
        else:
            return False
```

```python
# 557. Reverse Words in a String III

class Solution:
    def reverseWords(self, s: str) -> str:
        arr=list(s.split())
        res=[]
        for i in arr:
            res.append(i[::-1])
        return " ".join(res)
```

```python
# 917. Reverse Only Letters

class Solution:
    def reverseOnlyLetters(self, s: str) -> str:
        arr=[]
        temp=[]
        for i in s:
            if i.isalpha():
                arr.append(i)
            else:
                temp.append(i)
        arr=arr[::-1]
        i=0
        j=0
        k=0
        res=""
        while i<len(s):
            if s[i].isalpha():
                res+=arr[j]
                j+=1
            else:
                res+=temp[k]
                k+=1
            i+=1
        return res
```
